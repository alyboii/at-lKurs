# 📊 Train / Test / Validation — Final Öncesi Hızlı ama Derin Rehber
*(Veri Bilimci Perspektifiyle)*

---

## 1. Neden Veri Bölüyoruz?

Bir makine öğrenmesi modelinin **tek amacı**:
> Görmediği veride nasıl davranacağını ölçmek

Eğer model:
- Tüm veriyle **eğitilir**
- Aynı veriyle **test edilirse**

👉 Model **öğrenmez, ezberler**  
👉 Performans yapay olarak yüksek çıkar  
👉 Bu durum **overfitting** olarak adlandırılır

**Final yorumu:**  
> Test seti doğru ayrılmazsa modelin gerçek dünya performansı güvenilmez olur.

---

## 2. Train ve Test Setlerinin Rolleri

### 🔵 Train Set
- Model **burada öğrenir**
- Parametreler / ağırlıklar ayarlanır
- Model bu veriyi defalarca görür

### 🔴 Test Set
- Model **daha önce hiç görmemiştir**
- Sadece **en sonda** kullanılır
- Amaç: **tarafsız performans ölçümü**

⚠️ **Altın kural:**  
> Test seti, model geliştirme sürecine ASLA dahil edilmez.

---

## 3. Train/Test Oranı Değişirse Ne Olur?

### %95 Train / %5 Test
- Test seti çok küçük
- Performans ölçümü **oynak**
- Farklı çalıştırmalarda farklı sonuçlar

**Yorum:**  
> Test seti küçüldükçe ölçüm güvenilirliği azalır.

---

### %60 Train / %40 Test
- Ölçüm daha stabil
- Ama train küçüldüğü için model yeterince öğrenemeyebilir

**Denge:**  
> Train öğrenme için, test ölçüm içindir — ikisi de aşırı olmamalı

---

## 4. `random_state` Parametresi

- Verinin nasıl karıştırılacağını sabitler
- Aynı kod → aynı train/test split

❌ `random_state` yoksa:
- Her çalıştırmada farklı sonuç
- Deney tekrarlanamaz

**Final cümlesi:**  
> random_state, deneylerin tekrar edilebilirliğini sağlar.

---

## 5. Stratified Split (Çok Kritik)

Sınıf dengesizliği varsa (örneğin %90 / %10):

### ❌ Stratify Yoksa
- Azınlık sınıf test setine hiç düşmeyebilir
- Model her şeyi çoğunluk sınıf tahmin eder
- Accuracy yüksek ama model adaletsiz

### ✅ Stratify Varsa
- Train ve test’te sınıf oranları korunur
- Daha gerçekçi değerlendirme

**Final yorumu:**  
> Stratified split, sınıf dengesizliğinde zorunludur.

---

## 6. Validation Set Neden Var?

### Validation Set
- Hiperparametre ayarı
- Model seçimi
- Erken durdurma

### Test Set
- **Sadece final değerlendirme**

❌ Validation yokken test setiyle ayar yapılırsa:
- Test seti dolaylı validation olur
- Performans **iyimser** çıkar

**Ana fikir:**  
> Test seti ayar için değil, doğrulama içindir.

---

## 7. Cross-Validation (CV) Mantığı

Tek bir train/test split **şansa bağlıdır**.

Cross-validation:
- Veriyi K parçaya böler
- Her parça bir kez test olur
- Sonuçlar ortalanır

**Avantaj:** Daha güvenilir genelleme ölçümü  
**Dezavantaj:** Hesaplama maliyeti

---

## 8. K Değeri Değişirse Ne Olur?

### Küçük K (K=3)
- Hızlı
- Ölçüm kaba

### Büyük K (K=n → Leave-One-Out)
- Bias düşük
- Hesaplama maliyeti çok yüksek
- Varyans artabilir

**Final yorumu:**  
> K büyüdükçe değerlendirme detaylanır ama maliyet artar.

---

## 9. Zaman Serisi Veride Bölme

❌ Random split:
- Gelecek verisi geçmişi tahmin eder
- **Data leakage**

✅ Doğru yöntem:
- Geçmiş → Train
- Gelecek → Test

**Final cümlesi:**  
> Zaman bağımlı veride sıralı bölme yapılmalıdır.

---

## 10. Preprocessing Hatası (Yıldızlı)

Eğer:
- Scaling
- Normalizasyon
- Encoding

işlemleri **tüm veri üzerinde yapıp**
sonra CV uygularsan:

👉 Test bilgisi train’e sızar  
👉 Sonuçlar yanıltıcı olur

**Doğru çözüm:**  
> Preprocessing + model → pipeline  
> Pipeline, cross-validation içinde çalıştırılmalı

---

## 🎯 Final Öncesi 5 Maddelik Süper Özet

1. Test seti modele asla dokunmaz  
2. Stratify, sınıf dengesizliğinde şarttır  
3. random_state → tekrarlanabilirlik  
4. Validation ayar içindir, test doğrulama içindir  
5. Cross-validation tek split’ten daha güvenilirdir

---


# 🧠 Senaryo → Doğru Yöntem → Gerekçe Şablonu
*(Train / Test / Validation / Cross-Validation — Final Tipi)*

---

## 📌 Senaryo 1 — Sınıf Dengesizliği

### 🧩 Senaryo
Bir sınıflandırma probleminde veri setinin %90’ı sınıf A, %10’u sınıf B’den oluşmaktadır.  
Model train/test olarak rastgele (%80 / %20) bölünmüş ve yüksek accuracy elde edilmiştir.

---

### ✅ Doğru Yöntem
**Stratified Train/Test Split** kullanılmalıdır.

---

### 🧠 Gerekçe
Rastgele bölme, sınıf dengesizliği olan veri setlerinde azınlık sınıfın test veya train setine hiç düşmemesine yol açabilir.  
Bu durumda model, tüm örnekleri çoğunluk sınıf olarak tahmin ederek yüksek accuracy elde edebilir ancak gerçek performansı yansıtmaz.  
Stratified split, sınıf oranlarını train ve test setlerinde koruyarak daha adil ve gerçekçi bir değerlendirme sağlar.

---

## 📌 Senaryo 2 — Test Setiyle Model Ayarı

### 🧩 Senaryo
Bir model geliştirme sürecinde validation set ayrılmamış, farklı hiperparametreler denenirken test seti performansı sürekli kontrol edilmiştir.

---

### ✅ Doğru Yöntem
**Train / Validation / Test** ayrımı yapılmalıdır.

---

### 🧠 Gerekçe
Test seti model geliştirme sürecine dahil edilirse, test seti dolaylı olarak validation seti gibi kullanılmış olur.  
Bu durum test performansının iyimser çıkmasına ve modelin genelleme yeteneğinin yanlış değerlendirilmesine neden olur.  
Validation set, model ayarları için; test seti ise yalnızca nihai değerlendirme için kullanılmalıdır.

---

## 📌 Senaryo 3 — Küçük Veri Seti

### 🧩 Senaryo
Veri seti oldukça küçüktür ve tek bir train/test split ile elde edilen performansın güvenilir olup olmadığı sorgulanmaktadır.

---

### ✅ Doğru Yöntem
**K-Fold Cross-Validation** uygulanmalıdır.

---

### 🧠 Gerekçe
Tek bir train/test bölmesi şansa bağlı sonuçlar üretebilir.  
Cross-validation, veriyi birden fazla alt kümeye bölerek modelin farklı veri parçaları üzerindeki performansını ölçer.  
Bu sayede genelleme yeteneği daha güvenilir ve istikrarlı biçimde değerlendirilir.

---

## 📌 Senaryo 4 — Zaman Serisi Verisi

### 🧩 Senaryo
Bir zaman serisi veri setinde (tarihe bağlı ölçümler) veri rastgele train/test olarak bölünmüştür.

---

### ❌ Yanlış Yöntem
Random train/test split

---

### ✅ Doğru Yöntem
**Zaman temelli (sıralı) bölme**

---

### 🧠 Gerekçe
Zaman serisi verilerinde rastgele bölme, gelecekteki bilgilerin geçmişi tahmin etmek için kullanılmasına yol açar.  
Bu durum veri sızmasına (data leakage) neden olur ve model performansı gerçekçi olmaz.  
Doğru yaklaşım, geçmiş verilerin train, gelecekteki verilerin test olarak ayrılmasıdır.

---

## 📌 Senaryo 5 — Preprocessing ve Cross-Validation

### 🧩 Senaryo
Ölçekleme (scaling) işlemi tüm veri üzerinde uygulanmış, ardından K-Fold cross-validation yapılmıştır.

---

### ❌ Yanlış Yöntem
Preprocessing’in tüm veri üzerinde önceden yapılması

---

### ✅ Doğru Yöntem
**Pipeline + Cross-Validation**

---

### 🧠 Gerekçe
Preprocessing işlemleri tüm veri üzerinde yapılırsa, test fold’larındaki bilgi train fold’lara sızar.  
Bu da cross-validation sonuçlarının önyargılı ve iyimser çıkmasına neden olur.  
Pipeline kullanımı, preprocessing ve modelleme adımlarının her fold içinde doğru sırayla uygulanmasını sağlar.

---

## 🎯 Finalde Yazarken Kullanılacak Kalıp Cümleler

- “Bu yaklaşım veri sızmasına (data leakage) yol açar.”
- “Modelin genelleme yeteneği bu durumda doğru ölçülemez.”
- “Test seti, model geliştirme sürecine dahil edilmemelidir.”
- “Stratified bölme, sınıf dengesizliğinde daha sağlıklı sonuçlar verir.”
- “Cross-validation, tek bir train/test split’e göre daha güvenilir bir değerlendirme sunar.”

# 🧠 Veri Dönüştürme & Encoding — Yorumsal Senaryolar (Final Tipi)

---

## 📌 Senaryo 1 — Ölçeklendirme Yapılmadan Model Eğitimi

### 🧩 Senaryo
Bir veri setinde yaş (18–80), gelir (10.000–100.000) ve oda sayısı (1–5) değişkenleri
hiçbir ölçeklendirme yapılmadan k-NN algoritmasına verilmiştir.

---

### ❌ Yanlış Yöntem
Farklı ölçeklerdeki sayısal değişkenleri doğrudan modele vermek.

---

### ❓ Neden Yanlış?
Mesafe temelli algoritmalar (k-NN, k-means, SVM) değişkenler arasındaki **mutlak büyüklükleri**
kullanır.  
Bu durumda geniş aralığa sahip değişkenler (örneğin gelir), model tarafından
daha önemliymiş gibi algılanır.  
Sonuç olarak model, küçük ölçekli ama anlamlı değişkenleri ihmal eder.

---

### ✅ Nasıl Düzeltilir?
Veri seti **standardizasyon (Z-score)** veya **min-max normalizasyon** ile
aynı ölçeğe getirilmelidir.  
Bu sayede tüm değişkenler modele dengeli katkı sağlar.

---

## 📌 Senaryo 2 — Outlier İçeren Veride StandardScaler Kullanımı

### 🧩 Senaryo
Gelir değişkeninde uç değerler (outlier) bulunan bir veri setinde
StandardScaler kullanılmıştır.

---

### ❌ Yanlış Yöntem
Outlier içeren veride ortalama ve standart sapmaya dayalı ölçeklendirme kullanmak.

---

### ❓ Neden Yanlış?
StandardScaler, ortalama ve standart sapmayı kullandığı için
uç değerlerden ciddi biçimde etkilenir.  
Bu durum verinin büyük bölümünün sıkışmasına ve
ölçeklendirmenin anlamsızlaşmasına yol açar.

---

### ✅ Nasıl Düzeltilir?
**RobustScaler** kullanılmalıdır.  
RobustScaler, medyan ve IQR kullandığı için outlier’lara karşı daha dayanıklıdır.

---

## 📌 Senaryo 3 — Nominal Kategorik Değişkende Label Encoding

### 🧩 Senaryo
“Renk” değişkeni (Kırmızı, Mavi, Yeşil) Label Encoding ile 0, 1, 2 olarak kodlanmıştır.

---

### ❌ Yanlış Yöntem
Sırasız (nominal) kategorik değişkenlerde Label Encoding kullanmak.

---

### ❓ Neden Yanlış?
Label Encoding, sayısal bir sıralama varmış gibi algı oluşturur.  
Model, “2 > 1 > 0” şeklinde yapay bir büyüklük ilişkisi öğrenir.  
Bu ilişki gerçekte mevcut değildir ve modele yanlış bilgi yükler.

---

### ✅ Nasıl Düzeltilir?
Nominal kategorik değişkenler için **One-Hot Encoding / get_dummies** kullanılmalıdır.  
Bu yöntem kategoriler arasında yapay sıralama oluşturmaz.

---

## 📌 Senaryo 4 — Çok Fazla Kategoride One-Hot Encoding

### 🧩 Senaryo
Binlerce benzersiz kategori içeren bir değişken One-Hot Encoding ile dönüştürülmüştür.

---

### ❌ Yanlış Yöntem
Çok yüksek kardinaliteye sahip değişkenlerde One-Hot Encoding kullanmak.

---

### ❓ Neden Yanlış?
One-Hot Encoding, her kategori için ayrı bir sütun oluşturur.  
Bu durum:
- Boyut artışı (curse of dimensionality)
- Seyrek (sparse) veri
- Hesaplama ve bellek maliyeti

problemlerine yol açar.

---

### ✅ Nasıl Düzeltilir?
Duruma göre:
- Label Encoding
- Target / frequency encoding
- Embedding temelli yaklaşımlar
- Veya farklı model seçimi

tercih edilmelidir.

---

## 📌 Senaryo 5 — Curse of Dimensionality Sonrası KNN Kullanımı

### 🧩 Senaryo
One-Hot Encoding sonrası yüzlerce özelliğe sahip bir veri setinde
k-NN algoritması kullanılmıştır.

---

### ❌ Yanlış Yöntem
Yüksek boyutlu ve seyrek veride mesafe temelli algoritma kullanmak.

---

### ❓ Neden Yanlış?
Boyut arttıkça veri noktaları arasındaki mesafeler birbirine yaklaşır.  
Bu durumda “en yakın komşu” kavramı anlamını yitirir.  
Modelin genelleme yeteneği düşer.

---

### ✅ Nasıl Düzeltilir?
- Özellik seçimi
- Boyut indirgeme (PCA)
- Ağaç tabanlı modeller
- Regularization

gibi yöntemler tercih edilmelidir.

---

## 📌 Senaryo 6 — Encoding & Scaling CV Öncesi Yapılmış

### 🧩 Senaryo
Encoding ve scaling tüm veri üzerinde yapılmış, ardından cross-validation uygulanmıştır.

---

### ❌ Yanlış Yöntem
Ön işlemleri CV öncesinde tüm veri üzerinde gerçekleştirmek.

---

### ❓ Neden Yanlış?
Test fold’larındaki istatistiksel bilgi train fold’lara sızar.  
Bu durum **data leakage** oluşturur ve performans sonuçları iyimser çıkar.

---

### ✅ Nasıl Düzeltilir?
Encoding ve scaling işlemleri **Pipeline + ColumnTransformer**
kullanılarak cross-validation içine alınmalıdır.

---

## 🎯 Finalde Yazarken Kullanılacak Anahtar Cümleler

- “Bu yaklaşım modele yapay bir sıralama bilgisi yükler.”
- “Uç değerler nedeniyle ölçeklendirme yanıltıcı hâle gelir.”
- “Boyut artışı, mesafe temelli algoritmaların etkinliğini azaltır.”
- “Bu durum curse of dimensionality problemine yol açar.”
- “Daha sağlıklı bir değerlendirme için pipeline yapısı tercih edilmelidir.”

# 🧠 Makine Öğrenmesi — Genel Özet (Final Öncesi Tek Sayfa)

---

## 1️⃣ Veri Neden Bölünür? (Train / Validation / Test)

**Amaç:**  
Modelin **görmediği veride** nasıl davrandığını ölçmek.

- **Train:** Model öğrenir
- **Validation:** Model ayarları yapılır
- **Test:** Nihai, tarafsız değerlendirme

⚠️ **Altın kural:**  
> Test seti, model geliştirme sürecine ASLA dahil edilmez.

---

## 2️⃣ Train/Test Oranı Değişirse

- Test çok küçük → Ölçüm oynak
- Train çok küçük → Model yetersiz öğrenir

📌 **Denge gerekir** (%70–30, %80–20 yaygın)

---

## 3️⃣ random_state

- Split’i sabitler
- Deneylerin **tekrar edilebilir** olmasını sağlar

❌ Yoksa: Her çalıştırmada farklı sonuç

---

## 4️⃣ Stratified Split (Sınıf Dengesizliği)

- Azınlık sınıf varsa **zorunlu**
- Aksi halde model çoğunluk sınıfı ezberler

📌 **Final cümlesi:**  
> Stratified split, sınıf dağılımını korur ve adil değerlendirme sağlar.

---

## 5️⃣ Cross-Validation (CV)

**Neden?**  
Tek bir split şansa bağlıdır.

- K-Fold → Daha güvenilir performans
- K büyük → Daha detaylı ama pahalı

⚠️ Küçük veri setlerinde özellikle önemli

---

## 6️⃣ Zaman Serisi Verisi

❌ Random split → Gelecekten geçmişe bilgi sızar  
✅ Zaman sıralı split → Geçmiş train, gelecek test

📌 **Kavram:** Data Leakage

---

## 7️⃣ Ölçeklendirme (Scaling) — Ne Zaman Gerekli?

Gerekli olduğu durumlar:
- Mesafe temelli modeller (KNN, K-means, SVM)
- Gradient descent kullanan modeller
- Farklı ölçeklerde sayısal değişkenler

### Yöntemler:
- **StandardScaler:** Ortalama 0, std 1
- **Min-Max:** [0,1] aralığı
- **RobustScaler:** Outlier varsa

⚠️ Ağaç tabanlı modellerde genelde şart değil

---

## 8️⃣ Kategorik Değişkenler (Encoding)

### Nominal (sırasız):
- Renk, şehir, cinsiyet
- ✅ One-Hot Encoding
- ❌ Label Encoding

### Ordinal (sıralı):
- Küçük–orta–büyük
- ✅ Label / Ordinal Encoding

📌 **Yanlış encoding → modele sahte sıralama yükler**

---

## 9️⃣ One-Hot Sonrası Sorunlar

- Çok fazla sütun
- Sparse (seyrek) veri
- Curse of Dimensionality
- KNN / K-means performans düşüşü

### Çözümler:
- Özellik seçimi
- PCA / boyut indirgeme
- Regularization
- Ağaç tabanlı modeller
