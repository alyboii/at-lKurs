````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])
````

````python
param_grid = [
    {
        "model__penalty": ["l1"],
        "model__C": [0.01, 0.1, 1, 10],
        "model__solver": ["liblinear"],
        "model__class_weight": [None, "balanced"]
    },
    {
        "model__penalty": ["l2"],
        "model__C": [0.01, 0.1, 1, 10],
        "model__solver": ["liblinear", "lbfgs"],
        "model__class_weight": [None, "balanced"]
    }
]
````

````python
from sklearn.model_selection import GridSearchCV

grid_recall = GridSearchCV(
    pipeline,
    param_grid,
    scoring="recall",
    cv=5,
    n_jobs=-1
)

grid_recall.fit(X_train, y_train)
````

````python
grid_precision = GridSearchCV(
    pipeline,
    param_grid,
    scoring="precision",
    cv=5,
    n_jobs=-1
)

grid_precision.fit(X_train, y_train)
````

````python
grid_f1 = GridSearchCV(
    pipeline,
    param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1
)

grid_f1.fit(X_train, y_train)
````

````python
grid_auc = GridSearchCV(
    pipeline,
    param_grid,
    scoring="roc_auc",
    cv=5,
    n_jobs=-1
)

grid_auc.fit(X_train, y_train)
````

````python
from sklearn.metrics import (
    precision_score, recall_score, f1_score, roc_auc_score
)

def evaluate(model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:,1]
    
    return {
        "precision": precision_score(y_test, y_pred),
        "recall": recall_score(y_test, y_pred),
        "f1": f1_score(y_test, y_pred),
        "roc_auc": roc_auc_score(y_test, y_prob)
    }
````

````python
results = {
    "Recall Optimized": evaluate(grid_recall.best_estimator_, X_test, y_test),
    "Precision Optimized": evaluate(grid_precision.best_estimator_, X_test, y_test),
    "F1 Optimized": evaluate(grid_f1.best_estimator_, X_test, y_test),
    "ROC-AUC Optimized": evaluate(grid_auc.best_estimator_, X_test, y_test),
}

import pandas as pd
pd.DataFrame(results).T
````

## 2️⃣ Data Cleaning
The `TotalCharges` column was an "object" type and contained empty strings. We converted it to numeric format and handled missing values.

```python

# One-Hot Encoding for multi-category variables
